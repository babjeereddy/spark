```python
lines =["pyspark is powerful", "pyspark handles big data","big data is future"]
words_rdd =spark.sparkContext.parallelize(lines).flatMap(lambda lines : lines.split())

pair_rdd = words_rdd.map(lambda x : (x,1))

word_count =pair_rdd.reduceByKey(lambda x, y: x+y)
length_rdd =word_count.map(lambda x : (len(x[0]),x[1]))
length_rdd.collect()
group = length_rdd.reduceByKey(lambda x,y : x+y)
finalresult =group.mapValues(lambda x: f"Total words of this length: {x}").sortByKey(ascending=False)
finalresult.collect()
```
